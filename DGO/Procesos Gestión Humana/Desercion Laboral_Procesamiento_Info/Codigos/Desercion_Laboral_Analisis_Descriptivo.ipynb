## 2. Cargar información


```r
df <- read.csv(file.path("..", "Insumos", "datos_contratacion.csv"), sep = ";", stringsAsFactors = FALSE)
df$ESTADO <- as.factor(df$ESTADO)
names(df)
```

## 3. Análisis univariado (por ESTADO)

### 3.1 Estatus


```r
table(df$ESTADO) # Conteo
prop.table(table(df$ESTADO)) # Proporciones
```

### 3.2 Edad


```r
df %>% group_by(ESTADO) %>%
  summarise(
    media_edad = mean(EDAD, na.rm = TRUE),
    mediana_edad = median(EDAD, na.rm = TRUE),
    sd_edad = sd(EDAD, na.rm = TRUE)
  )
```

### 3.3 Salario anual


```r
df %>% group_by(ESTADO) %>%
  summarise(
    media_salario = mean(SALARIO_ANUAL, na.rm = TRUE),
    mediana_salario = median(SALARIO_ANUAL, na.rm = TRUE)
  )
```

### 3.4 Salario por hora


```r
df %>% group_by(ESTADO) %>%
  summarise(
    media_hrly = mean(SALARIO_POR_HORA, na.rm = TRUE),
    mediana_hrly = median(SALARIO_POR_HORA, na.rm = TRUE)
  )
```

### 3.5 Satisfacción laboral


```r
df %>% group_by(ESTADO, SATISFACCION_LABORAL) %>%
  summarise(n = n()) %>% mutate(prop = n / sum(n))
```

### 3.6 Estado civil


```r
df %>% group_by(ESTADO, ESTADO_CIVIL) %>%
  summarise(n = n()) %>% mutate(prop = n / sum(n))
```

### 3.7 Código de trabajo


```r
df %>% group_by(ESTADO, CODIGO_TRABAJO) %>%
  summarise(n = n()) %>% mutate(prop = n / sum(n))
```

### 3.8 Grupo de trabajo


```r
df %>% group_by(ESTADO, GRUPO_TRABAJO) %>%
  summarise(n = n()) %>% mutate(prop = n / sum(n))
```

### 3.9 Evaluación de desempeño


```r
df %>% group_by(ESTADO, EVALUACION_DESEMPENO) %>%
  summarise(n = n()) %>% mutate(prop = n / sum(n))
```

### 3.10 Viajes requeridos


```r
df %>% group_by(ESTADO, VIAJES_REQUERIDOS) %>%
  summarise(n = n()) %>% mutate(prop = n / sum(n))
```
